```python
# Core imports for the whole lab
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')      # symbolic variables we'll reuse
sp.init_printing()            # pretty-print symbolic math
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
expr = x**2
deriv = sp.diff(expr, x)          # differentiate w.r.t. x
print('d/dx (x**2) =', deriv)     # 2*x

# Evaluate the symbolic derivative at x = 3
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
def g(x):
    return x**3 + 2*x

import sympy as sp


def g_num(x):
    return x**3 + 2*x


x_val = 2
h = 1e-5
numerical_derivative = (g_num(x_val + h) - g_num(x_val - h)) / (2 * h)

print(f"1. Numerical Derivative at x = {x_val}: {numerical_derivative}")


x = sp.Symbol('x')
g_sym = x**3 + 2*x
symbolic_derivative_expr = sp.diff(g_sym, x)

print(f"2. Symbolic Derivative Expression: {symbolic_derivative_expr}")

evaluated_symbolic = float(symbolic_derivative_expr.subs(x, x_val))

print(f"3. Evaluated Symbolic Derivative at x = {x_val}: {evaluated_symbolic}")
print(f"   Matches Step 1? {abs(numerical_derivative - evaluated_symbolic) < 1e-5}")
```
