```python
import sympy as sp

def g(x):
    return x**3 + 2*x

# 1. Numerical derivative at x = 2 (use h = 1e-6)
h = 1e-6
x = 2

numerical_derivative = (g(x + h) - g(x)) / h

print("Numerical derivative at x = 2:", numerical_derivative)

# 2. Symbolic derivative of x**3 + 2*x
x_sym = sp.Symbol('x')
expr = x_sym**3 + 2*x_sym

symbolic_derivative = sp.diff(expr, x_sym)

print("Symbolic derivative:", symbolic_derivative)

# 3. Evaluate the symbolic derivative at x = 2 and compare
symbolic_at_2 = symbolic_derivative.subs(x_sym, 2)

print("Symbolic derivative at x = 2:", symbolic_at_2)
print("Numerical and symbolic match:", abs(numerical_derivative - float(symbolic_at_2)) < 1e-5)
```

```python
import sympy as sp

x, y = sp.symbols('x y')


#  2A. PARTIAL DERIVATIVES


f2 = x**2 + 3*x*y + y**2

# Partial derivative with respect to x
print('df/dx =', sp.diff(f2, x))

# Partial derivative with respect to y
print('df/dy =', sp.diff(f2, y))


#  2B. THE GRADIENT (vector of partials)


# The gradient contains all partial derivatives
grad = [sp.diff(f2, v) for v in (x, y)]

print('grad f =', grad)

# Evaluate the gradient at (x=1, y=2)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]

print('grad f at (1, 2) =', grad_at)
```

```python
h2 = x**2 * y + sp.sin(y)

# 1. dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)

# 2. Assemble the gradient list
grad = [dh_dx, dh_dy]

print("Gradient =", grad)

# 3. Evaluate at (x=2, y=0)
grad_at = [g.subs({x: 2, y: 0}) for g in grad]

print("Gradient at (2, 0) =", grad_at)
```

```python
import sympy as sp

x = sp.Symbol('x')

# 3A. CHAIN RULE BY HAND vs SymPy

# y = sin(x**2)

# Chain rule:
# dy/dx = cos(x**2) * 2*x

by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)

print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
# 3B. CHAINING THREE FUNCTIONS


import sympy as sp

x = sp.Symbol('x')

# y = (3x + 1)^4
expr3 = (3*x + 1)**4

# Differentiate using SymPy
derivative = sp.diff(expr3, x)

print('d/dx (3x+1)^4 =', derivative)
```

```python
# y = exp(x**2 + 1)
# inner = x**2 + 1
# inner' = 2x
# outer' = exp(inner)

# 1. By hand (as a SymPy expression)
by_hand = sp.exp(x**2 + 1) * 2*x

# 2. With sp.diff
by_sympy = sp.diff(sp.exp(x**2 + 1), x)

# 3. Confirm they match
print("By hand :", by_hand)
print("By SymPy:", by_sympy)
print("Match?  ", sp.simplify(by_hand - by_sympy) == 0)
```
