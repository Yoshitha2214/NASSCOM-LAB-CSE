```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create the dataset
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5",
                "S6", "S7", "S8", "S9", "S10"],

    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],

    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],

    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],

    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],

    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}


df = pd.DataFrame(data)


print("Complete Dataset:")
print(df)

print("\nNumber of students:", len(df))


print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)


print("\nFirst five records:")
print(df.head())

print("\nLast five records:")
print(df.tail())


print("\nDataFrame Information:")
df.info()
```

pandas


```python

df = pd.DataFrame(data)


print("Complete Dataset:")
print(df)


print("\nAverage values:")
print(df[[
    "Study_Hours",
    "Attendance",
    "Assignment_Score",
    "Previous_Score",
    "Final_Score"
]].mean())

highest = df.loc[df["Final_Score"].idxmax()]
print("\nStudent with highest Final Score:")
print(highest)


lowest = df.loc[df["Final_Score"].idxmin()]
print("\nStudent with lowest Final Score:")
print(lowest)

print("\nStudents with Attendance greater than 80%:")
print(df[df["Attendance"] > 80])


print("\nStudents with Final Score greater than 80:")
print(df[df["Final_Score"] > 80])


print("\nStatistical Summary:")
print(df.describe())
```

numpy

```python

final_scores = df["Final_Score"].to_numpy()

print("Final Scores:")
print(final_scores)


mean_score = np.mean(final_scores)
median_score = np.median(final_scores)
maximum_score = np.max(final_scores)
minimum_score = np.min(final_scores)
std_deviation = np.std(final_scores)

print("\nMean:", mean_score)
print("Median:", median_score)
print("Maximum:", maximum_score)
print("Minimum:", minimum_score)
print("Standard Deviation:", std_deviation)
```

linear algebra

```python

X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values

print("Matrix X:")
print(X)


print("\nShape of X:")
print(X.shape)

print("\nTranspose of X:")
print(X.T)

print("\nX transpose X:")
print(X.T @ X)

study_hours = df["Study_Hours"].values
final_scores = df["Final_Score"].values

dot_product = np.dot(study_hours, final_scores)

print("\nDot product between Study Hours and Final Score:")
print(dot_product)
```

claculus:differentiation function

```python

def F(x):
    return 40 + 5*x - 0.2*x**2

def dF(x):
    return 5 - 0.4*x

for x in range(1, 11):
    print(x, F(x), dF(x))
```

calculus:integration function

```python
import numpy as np

def L(x):
    return 5*x - 0.2*x**2

x = np.linspace(0, 5, 1000)

y = L(x)

area = np.trapezoid(y, x)

print("Numerical approximation:", area)
```

linear algebra

```python
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values

X = np.column_stack((np.ones(len(X)), X))

beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
```
