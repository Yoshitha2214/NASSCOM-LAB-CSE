```python
import numpy as np

# ---------------------------
# Task 1 - Python Basics
# ---------------------------

time = [0, 1, 2, 3, 4, 5]
temperature = [32, 35, 38, 40, 42, 45]
voltage = [400, 395, 390, 385, 380, 375]
current = [20, 22, 25, 28, 30, 32]
charge = [95, 88, 80, 70, 60, 48]

print("Battery Sensor Readings")
for i in range(len(time)):
    print(f"Time={time[i]}h, Temp={temperature[i]}°C, Voltage={voltage[i]}V, Current={current[i]}A, Charge={charge[i]}%")

print("\nTemperature Status:")
for t in temperature:
    if t > 40:
        print("High Temperature Alert")
    else:
        print("Temperature Normal")

print("\nBattery Charge Status:")
for c in charge:
    if c < 50:
        print("Low Battery Alert")
    else:
        print("Battery Charge Normal")

def battery_status(charge):
    if charge > 70:
        return "High"
    elif charge >= 50:
        return "Medium"
    else:
        return "Low"

print("\nBattery Status:")
for c in charge:
    print(f"Charge={c}% --> {battery_status(c)}")

# ---------------------------
# Task 2 - NumPy Analysis
# ---------------------------

D = np.array([
    [32, 400, 20, 95],
    [35, 395, 22, 88],
    [38, 390, 25, 80],
    [40, 385, 28, 70],
    [42, 380, 30, 60],
    [45, 375, 32, 48]
])

print("\nNumPy Analysis")

print("Shape of Array:", D.shape)

avg_values = np.mean(D, axis=0)
print("Average Temperature =", avg_values[0])
print("Average Voltage =", avg_values[1])
print("Average Current =", avg_values[2])
print("Average Charge =", avg_values[3])

print("Maximum Temperature =", np.max(D[:, 0]))
print("Minimum Temperature =", np.min(D[:, 0]))

print("\nVoltage Column:")
print(D[:, 1])

print("\nRows with Temperature > 40°C")
print(D[D[:, 0] > 40])

print("\nRows with Charge < 50%")
print(D[D[:, 3] < 50])

# ---------------------------
# Task 3 - Linear Algebra
# ---------------------------

A = np.array([
    [2, 1],
    [1, 3]
])

X = np.array([
    [40],
    [25]
])

Y = np.dot(A, X)

print("\nLinear Algebra")

print("Y = A * X")
print(Y)

print("\nTranspose of A:")
print(A.T)

print("\nDeterminant of A:")
print(np.linalg.det(A))

print("\nInverse of A:")
print(np.linalg.inv(A))

B = np.array([
    [100],
    [120]
])

solution = np.linalg.solve(A, B)

print("\nSolution of AX = B")
print("x =", solution[0][0])
print("y =", solution[1][0])

# ---------------------------
# Task 4 - Calculus
# ---------------------------

print("\nCalculus")

def dC_dt(t):
    return -2 - t

print("dC/dt = -2 - t")

print("Discharge Rate at t=2:", dC_dt(2))
print("Discharge Rate at t=5:", dC_dt(5))

if abs(dC_dt(5)) > abs(dC_dt(2)):
    print("Battery is losing charge at an increasing rate.")

# ---------------------------
# Final Analysis
# ---------------------------

print("\nFinal Analysis")

for i in range(len(temperature)):
    if temperature[i] > 40:
        print("Temperature first exceeds 40°C at time =", time[i], "hours")
        break

charge_loss = charge[0] - charge[-1]
print("Total Charge Loss =", charge_loss, "%")

print("Average Temperature =", avg_values[0])
print("Average Voltage =", avg_values[1])
print("Average Current =", avg_values[2])
print("Average Charge =", avg_values[3])

print("Transformed Output Y =")
print(Y)

print("x =", solution[0][0], ", y =", solution[1][0])

print("\nFinal Decision: MONITOR")
```
