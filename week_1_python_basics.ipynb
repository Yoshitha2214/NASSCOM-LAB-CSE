function

```python
def greet():
  print("Hello world")
greet()
```

function with parameters

```python
def greet(name):
  print(f"Name:{name}")
greet("Yoshitha")
greet("Pranavi")
```

function with default parameters

```python
def greet(name="Hello"):
  print(f"Name:{name}")
greet("Yoshitha")
greet("Pranavi")
greet()
```

function using *args

```python
def greet(*names):
  print("Names passes: ",names)
  for name in names:
    print("Hello",name)
greet("Yoshitha","Sharanya","Pranavi")
```

function using **kwargs

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Yoshitha", age=25, city="Hyderabad")
```

function using *args and **kwargs

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Yoshitha", city="Hyd")
```

function nesting

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```
